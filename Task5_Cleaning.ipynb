# Task 5 – Data Cleaning using Pandas
Titanic Dataset Cleaning in Google Colab/Jupyter

## Step 1 – Import Libraries

```python
import pandas as pd
import numpy as np
```

## Step 2 – Load Dataset

```python
df = pd.read_csv('titanic_sample.csv')
df.head()
```

## Step 3 – Check Structure & Missing Values

```python
df.info()
df.isnull().sum()
```

## Step 4 – Remove Duplicates

```python
df = df.drop_duplicates()
```

## Step 5 – Handle Missing Values
- Age → median
- Fare → mean
- Embarked → mode

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Fare'] = df['Fare'].fillna(df['Fare'].mean())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

## Step 6 – Create New Columns (Feature Engineering)

```python
df['Age_Group'] = pd.cut(df['Age'], bins=[0,18,35,60], labels=['Child','Adult','Senior'])
df['Fare_Band'] = pd.cut(df['Fare'], bins=3, labels=['Low','Medium','High'])
```

## Step 7 – Save Cleaned Data

```python
df.to_csv('cleaned_data.csv', index=False)
```

## Step 8 – Summary
Data cleaned, transformed, and ready for analysis or modeling.
